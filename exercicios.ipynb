```python
from sklearn.linear_model import LinearRegression


modelo = LinearRegression()

duracoes = [
    [10],
    [15],
    [20],
    [25],
    [30],
    [35],
    [40],
    [50],
    [55],
    [60]
]
precos = [
    10.00,
    13.00,
    14.00,
    18.00,
    19.00,
    23.00,
    24.00,
    30.00,
    32.00,
    34.00
]

modelo.fit(duracoes, precos)


preco_45 = modelo.predict([[45]])
preco_20 = modelo.predict([[20]])
preco_60 = modelo.predict([[60]])

print(f"45 minutos: R$ {preco_45[0]:.2f}")
print(f"20 minutos: R$ {preco_20[0]:.2f}")
print(f"60 minutos: R$ {preco_60[0]:.2f}")

```
